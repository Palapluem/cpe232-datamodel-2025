Include appropriate chart titles and percentage labels.

```python
# Write your code here
age_group_pie = df_remove_rows['age_group'].value_counts()
bmi_category_pie = df_remove_rows['bmi_category'].value_counts()

fig, (ax1, ax2)= plt.subplots(1, 2, figsize=(15, 6))
ax1.pie(age_group_pie.values, labels=age_group_pie.index, autopct='%1.1f%%',shadow=True, startangle=90)
ax1.set_title("The distribution of customers with Age group")
ax2.pie(bmi_category_pie.values, labels=bmi_category_pie.index, autopct='%1.1f%%',shadow=True, startangle=90)
ax2.set_title("The distribution of customers with BMI categories")
plt.show()
```
