#### 3.3 Replace missing values with average (Total points = 21)

[3 points] Create a new dataframe `df_new` which contains only these attributes: `Rooms`, `Price`, `Distance`, `BuildingArea`. Then, likewise, display the first 5 rows.

```python
# Write your code here
df_new = df[["Rooms", "Price", "Distance", "BuildingArea"]]
df_new.head()
```

We are going to replace missing values in the attribute `BuildingArea` with its average.

[3 points] First, compute average value of the attribute `BuildingArea`.

```python
# Write your code here
average_building_area = df_new['BuildingArea'].mean()
print(f"Average value of hte attribute BuildingArea: {average_building_area}")
```

```python
# Write your code here
df_new['BuildingArea'] = df_new['BuildingArea'].fillna(average_building_area)
print(df_new.head(10))
```

```python
# Write your code here
std_dev_after = df_new['BuildingArea'].std()
print(f"A standard deviation of `BuildingArea` after the data imputation: {std_dev_after}")
```

```python
# Write your code here
std_dev_before = df['BuildingArea'].std()
std_dev_different = std_dev_before - std_dev_after
print(f"The difference of the BuildingArea standard deviation before versus after replacing missing values with its mean value: {std_dev_different}")
```

### Subtask #4: Create some visualizations

```python
import matplotlib.pyplot as plt
```

##### 4.1 Investigate types of rooms (Total points = 8)

[5 points] *Use the original dataframe* `df`, create a *horizontal* bar chart to present the attributes `Rooms`. Also display the chart title and data labels.

```python
# Write your code here
df["Rooms"].value_counts().plot(kind='barh', figsize=(6, 5))
plt.title('Distribution of Rooms')
plt.xlabel('Number of Properties')
plt.ylabel('Number of Rooms')
plt.show()
```

[3 points] Briefly describe this visualization in your own words, and does this information help with the decision making?

**ANS:** กราฟนี้แสดงการกระจายตัวของจำนวนห้อง (Rooms) ในตลาดอสังหาริมทรัพย์ โดยพบว่าบ้านส่วนใหญ่มีจำนวน 2-3 ห้อง (เป็นแท่งที่ยาวที่สุด) และบ้านที่มีจำนวนห้องน้อยมากหรือมากเกินไปจะมีจำนวนประกาศขายน้อย ข้อมูลนี้ ช่วยในการตัดสินใจ ได้อย่างมากสำหรับผู้พัฒนาหรือผู้ลงทุน เพื่อให้ทราบว่าขนาดบ้านแบบใดที่เป็นที่ต้องการของตลาดมากที่สุด ทำให้สามารถวางแผนการสร้างหรือการซื้อเพื่อเก็งกำไรได้ตรงตามความต้องการของลูกค้าส่วนใหญ่ครับ

##### 4.2 Investigate types of methods (Total points = 8)

[5 points] *Use the original dataframe* `df`, create a pie chart to present the attributes `Method`. Also display the chart title and data labels.

```python
# Write your code here
df["Method"].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(6, 5))
plt.title('Distribution of Method')
plt.show()
```

[3 points] Briefly describe this visualization in your own words, and does this information help with the decision making?

**ANS:** กราฟวงกลมนี้แสดงสัดส่วนเปอร์เซ็นต์ของวิธีการขายบ้าน (Method) ในตลาด เช่น การขายสำเร็จ (S), การขายก่อนการประมูล (SP), หรือการประมูลไม่สำเร็จ (PI) ข้อมูลนี้ ช่วยในการตัดสินใจ ได้ว่าสภาพตลาดในขณะนั้นเป็นอย่างไร หากมีสัดส่วนการขายสำเร็จ (S) สูง แสดงว่าตลาดมีความต้องการซื้อมาก แต่หากมีสัดส่วนบ้านที่ประมูลไม่สำเร็จ (PI หรือ VB) สูง ผู้ขายอาจต้องพิจารณาปรับราคามือตั้งต้น หรือเลือกวิธีการขายใหม่ให้เหมาะสมกับสถานการณ์ครับ

##### 4.3 Investigate council areas (Total points = 8)

[5 points] *Use the original dataframe* `df`, create a bar chart to present the attributes `CouncilArea`. Also display the chart title and data labels.

```python
# Write your code here
df["CouncilArea"].value_counts().plot(kind='bar', figsize=(10, 4))
plt.title('Distribution of CouncilArea')
plt.xlabel("Council Area")
plt.ylabel("Number of Properties")
plt.xticks(rotation = 45, ha = 'right')
plt.show()
```

[3 points] Briefly describe this visualization in your own words, and does this information help with the decision making?

**ANS:** กราฟนี้แสดงจำนวนรายการขายอสังหาริมทรัพย์แบ่งตามเขตการปกครอง (Council Area) โดยพบว่าบางเขต เช่น Moreland หรือ Boroondara มีจำนวนบ้านประกาศขายสูงมาก ข้อมูลนี้ ช่วยในการตัดสินใจ สำหรับผู้ซื้อหรือนักลงทุนเพื่อเลือกทำเลที่มีตัวเลือก (Supply) หลากหลาย และแสดงถึงทำเลที่มีการหมุนเวียนของตลาดสูง ซึ่งมักจะเป็นพื้นที่ที่มีสิ่งอำนวยความสะดวกครบถ้วนและเป็นที่ต้องการ ทำให้ง่ายต่อการเปรียบเทียบราคาและตัดสินใจซื้อครับ
