##### 4.1 Investigate impact of project (Total points = 12)

[5 points] Using the dataframe `df_remove_rows`, create a bar chart showing the average salary for each project count.

The x-axis should represent the number of projects, and the y-axis should represent the average salary.

Include an appropriate plot title and axis labels.

```python
# Write your code here
project_salary = df_remove_rows.groupby('Projects_Count')['Salary_Offered_USD'].mean()
plt.figure(figsize=(10, 6))
project_salary.plot(kind='bar', color='skyblue')
plt.title("Average of salary for each project count")
plt.xlabel("Number of Projects")
plt.ylabel("Average of Salary (USD)")
plt.show()
```

##### 4.2 Investigate proportion of students in different degrees via pie charts (Total points = 12)

[5 points] Using the dataframe `df_remove_rows`, create two pie charts showing the distribution of degree categories: one for students who received placement offers and one for students who did not.

Include appropriate chart titles and percentage labels.

```python
# Write your code here
placement_yes = df_remove_rows[df_remove_rows['Placement_Offer'] == 'Yes']['Degree'].value_counts()
placement_no = df_remove_rows[df_remove_rows['Placement_Offer'] == 'No']['Degree'].value_counts()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 7))
ax1.pie(placement_yes, labels=placement_yes.index, autopct='%1.1f%%')
ax1.set_title('Degree for students with received placement offers')
ax2.pie(placement_no, labels=placement_no.index, autopct='%1.1f%%')
ax2.set_title('Degree for students without received placement offers')
plt.show()
```

[7 points] Interpret the visualizations above. What insights can you draw from them? Provide a reasoned explanation and discuss any potential limitations of your interpretation.

**ANS:** แผนภูมิวงกลมแสดงให้เห็นว่านักศึกษาที่ได้รับงาน (Yes) และไม่ได้งาน (No) มีการกระจายตัวของสาขาวิชาที่ใกล้เคียงกันมาก คือทุกสาขามีโอกาสได้งานพอๆ กัน **ข้อจำกัด (Limitations):** การดู Pie Chart แยกกันสองวงทำให้นำข้อมูลมาเปรียบเทียบความแตกต่างเล็กน้อยได้ยากครับ

##### 4.3 Investigate proportion of students in different degrees via a bar chart  (Total points = 16)

[5 points] Using the dataframe `df_remove_rows`, create a side-by-side bar chart to compare the proportion of students across degree categories by placement status (Placement offer: Yes vs No).

The x-axis should represent degree categories, and the y-axis should represent the proportion of students.

Include appropriate chart titles, axis labels, and a legend.

```python
# Write your code here
degree_placement = df_remove_rows.groupby(['Degree', 'Placement_Offer']).size().unstack()
degree_placement.plot(kind='bar', figsize=(10, 5))
plt.title("Placement Status by Degree")
plt.xlabel("Degree")
plt.ylabel("Number of Students")
plt.legend(title="Offer Received")
plt.xticks(rotation=45)
plt.show()
```

[7 points] Interpret the visualization above. What insights can you draw from it? Provide a reasoned explanation and discuss any potential limitations of your interpretation.

**ANS:** แผนภูมิแท่งคู่นี้ชัดเจนกว่าวงกลม เพราะเราสามารถเห็น "จำนวนคน" จริงๆ ของแต่ละสาขาที่ได้งานและไม่ได้งานเปรียบเทียบกันได้ทันที พบว่าโครงสร้างการจ้างงานมีความใกล้เคียงกันในทุกสาขา
**ข้อจำกัด (Limitations):** กราฟนี้บอกเพียงปริมาณ แต่ไม่ได้บอกถึงเหตุผลเบื้องหลังว่าทำไมคนกลุ่มนั้นถึงไม่ได้งาน เช่น อาจจะมาจากคะแนนทักษะอื่นๆ ครับ
