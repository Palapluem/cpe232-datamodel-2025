#### 3.2 สร้างคอลัมน์ใหม่ (Total points = 23)

[5 points] Create a new column and name it `major_expense`. This column contains values that are the sum of housing, food, and transportation.

สร้างคอลัมน์ใหม่ชื่อ `major_expense` โดยมีค่าที่ได้จากผลรวมของค่าใช้จ่ายด้านที่อยู่อาศัย อาหาร และการเดินทาง

````python
# Write your code here
df['major_expense'] = df['housing'] + df['food'] + df['transportation']
df.head()
````

[8 points] Create another new column and name it `major_expense_ratio` which is based on the following formula:

สร้างคอลัมน์ใหม่ชื่อ `major_expense_ratio` โดยอิงจากสูตรการคำนวณต่อไปนี้

```
major_expense_ratio = (major_expense * 100) / monthly_income
```

````python
# Write your code here
df['major_expense_ratio'] = (df['major_expense'] * 100) / df['monthly_income']
df.head()
````

[10 points] According to the results in previous cell(s), do most students experience financial difficulties as a result of exceeding their monthly income? What is the percentage of those who experience financial difficulties and those who do not? Show your work and and analysis below.

จากผลลัพธ์ในเซลล์ก่อนหน้า นักศึกษาส่วนใหญ่ประสบปัญหาทางการเงินเนื่องจากใช้จ่ายเกินรายได้ต่อเดือนหรือไม่? คำนวณเปอร์เซ็นต์ของนักศึกษาที่ประสบปัญหาทางการเงินและนักศึกษาที่ไม่ประสบปัญหา พร้อมบรรยายผลการวิเคราะห์

**ANS:**  จากการวิเคราะห์โดยเปรียบเทียบรายจ่ายรวมทุกหมวดกับรายได้ต่อเดือน พบว่า นักศึกษาส่วนใหญ่ "ประสบปัญหาทางการเงิน" เนื่องจากมีการใช้จ่ายเกินรายได้ที่ได้รับ โดยมีสัดส่วนนักศึกษาที่ประสบปัญหาทางการเงิน (ใช้จ่ายเกินรายได้) คิดเป็น 53.6 % และนักศึกษาที่ไม่ประสบปัญหาทางการเงินคิดเป็น 40.4 % ครับ



````python
# Write your code here
over100_count = df[df['major_expense_ratio'] >= 100].shape[0]
under100_count = df[df['major_expense_ratio'] < 100].shape[0]

total = len(df)
percent_over = (over100_count / total) * 100
percent_under = (under100_count / total) * 100

print(f"เปอร์เซ็นต์ของนักศึกษาที่ประสบปัญหาทางการเงิน: {percent_over}")
print(f"เปอร์เซ็นต์ของนักศึกษาที่ไม่ประสบปัญหาทางการเงิน: {percent_under}")
````

##### 4.1 วิเคราะห์สาขาวิชาต่างๆ (Total points = 10)

สร้าง แผนภูมิวงกลม (pie chart) เพื่อแสดงค่าที่ไม่ซ้ำกันของคุณลักษณะ ```major``` ในการแสดงผล ให้แสดงชื่อแผนภูมิ, เปอร์เซ็นต์การกระจาย, และคำอธิบายสัญลักษณ์ (legend)

````python
# Write your code here
major_counts = df['major'].value_counts()
unique_majors = df['major'].unique()

plt.figure(figsize=(10, 6))
plt.pie(major_counts, labels=major_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Distibution of Student by major')
plt.legend(unique_majors, title="Majors", loc="lower left", bbox_to_anchor = (-0.3, 0))
plt.show()
````

### Subtask #5: จัดกลุ่มข้อมูล
